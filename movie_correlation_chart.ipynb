```python
#Storing the equation above as a variable in the correlation_matrix value
correlation_matrix = df.corr( numeric_only='True') 

#Making a Seaborn heatmap using the correlation matrix I made above
#Annot is for writing corresponding data values in each cell
sns.heatmap(correlation_matrix, annot=True)

plt.title('Correlation Matric for Numeric Features')

plt.xlabel('Movie Features')

plt.ylabel('Movie Features')

#Now to see the results
plt.show()
```

```python
df_numerize = df
for col_name in df_numerize.columns:
    if(df_numerize[col_name].dtype == 'object'):
        df_numerize[col_name] = df_numerize[col_name].astype('category')
        df_numerize[col_name] = df_numerize[col_name].cat.codes
df_numerize
```

```python
df_numerize.corr()
```

```python
correlation_mat = df_numerize.corr()

corr_pairs = correlation_mat.unstack()

corr_pairs
```

```python
sorted_pairs = corr_pairs.sort_values()

sorted_pairs
```

```python
high_corr = sorted_pairs[(sorted_pairs) > .5]

high_corr
```
